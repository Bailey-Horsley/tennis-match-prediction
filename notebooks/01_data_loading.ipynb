```python
import pandas as pd
from pathlib import Path

# -----------------------------------
# 1. Define data paths
# -----------------------------------

DATA_DIR = Path("../data")
OUTPUT_DIR = DATA_DIR / "combined"
OUTPUT_DIR.mkdir(exist_ok=True)

print("=" * 60)
print("DATA LOADING CHECK")
print("=" * 60)

print("\nCurrent working directory:")
print(Path.cwd())

print("\nData directory exists:", DATA_DIR.exists())
print("Data directory path:", DATA_DIR.resolve())

print("\nFiles in data directory:")
for f in sorted(DATA_DIR.iterdir()):
    print(" -", f.name)

# -----------------------------------
# 2. Locate files
# -----------------------------------

match_files = sorted(DATA_DIR.glob("atp_matches_*.csv"))

ranking_files = [
    DATA_DIR / "atp_rankings_10s.csv",
    DATA_DIR / "atp_rankings_20s.csv",
    DATA_DIR / "atp_rankings_current.csv",
]

players_file = DATA_DIR / "atp_players.csv"

print("\n" + "=" * 60)
print("FILES FOUND")
print("=" * 60)

print(f"\nFound {len(match_files)} match files:")
for f in match_files:
    print(" -", f.name)

print("\nRanking files:")
for f in ranking_files:
    print(" -", f.name, "| exists:", f.exists())

print("\nPlayers file:")
print(" -", players_file.name, "| exists:", players_file.exists())

# -----------------------------------
# 3. Load match files
# -----------------------------------

print("\n" + "=" * 60)
print("LOADING MATCH DATA")
print("=" * 60)

df_matches = pd.concat(
    [pd.read_csv(f) for f in match_files],
    ignore_index=True
)

# Parse tourney_date
df_matches["tourney_date"] = pd.to_datetime(
    df_matches["tourney_date"],
    format="%Y%m%d",
    errors="coerce"
)

print("\nMatches loaded successfully.")
print("Shape:", df_matches.shape)

print("\nFirst 5 rows of match data:")
print(df_matches.head())

print("\nMatch columns:")
for col in df_matches.columns:
    print(col)

print("\nYears in match dataset:")
print(sorted(df_matches["tourney_date"].dt.year.dropna().unique()))

print("\nTop 15 missing-value counts in match data:")
print(df_matches.isnull().sum().sort_values(ascending=False).head(15))

# -----------------------------------
# 4. Load ranking files
# -----------------------------------

print("\n" + "=" * 60)
print("LOADING RANKING DATA")
print("=" * 60)

df_rankings = pd.concat(
    [pd.read_csv(f) for f in ranking_files if f.exists()],
    ignore_index=True
)

# Parse ranking_date
df_rankings["ranking_date"] = pd.to_datetime(
    df_rankings["ranking_date"],
    format="%Y%m%d",
    errors="coerce"
)

print("\nRankings loaded successfully.")
print("Shape:", df_rankings.shape)

print("\nFirst 5 rows of ranking data:")
print(df_rankings.head())

print("\nRanking columns:")
for col in df_rankings.columns:
    print(col)

print("\nYears in ranking dataset:")
print(sorted(df_rankings["ranking_date"].dt.year.dropna().unique()))

print("\nTop 15 missing-value counts in ranking data:")
print(df_rankings.isnull().sum().sort_values(ascending=False).head(15))

# -----------------------------------
# 5. Load player file
# -----------------------------------

print("\n" + "=" * 60)
print("LOADING PLAYER DATA")
print("=" * 60)

df_players = pd.read_csv(players_file)

# Parse date of birth
df_players["dob"] = pd.to_datetime(
    df_players["dob"],
    format="%Y%m%d",
    errors="coerce"
)

print("\nPlayers loaded successfully.")
print("Shape:", df_players.shape)

print("\nFirst 5 rows of player data:")
print(df_players.head())

print("\nPlayer columns:")
for col in df_players.columns:
    print(col)

print("\nTop 15 missing-value counts in player data:")
print(df_players.isnull().sum().sort_values(ascending=False).head(15))

# -----------------------------------
# 6. Quick sanity checks
# -----------------------------------

print("\n" + "=" * 60)
print("SANITY CHECKS")
print("=" * 60)

winner_ids = set(df_matches["winner_id"].dropna().unique())
loser_ids = set(df_matches["loser_id"].dropna().unique())
match_player_ids = winner_ids.union(loser_ids)
player_file_ids = set(df_players["player_id"].dropna().unique())

common_ids = match_player_ids.intersection(player_file_ids)

print(f"\nUnique player IDs in matches: {len(match_player_ids):,}")
print(f"Unique player IDs in player file: {len(player_file_ids):,}")
print(f"Matching player IDs: {len(common_ids):,}")

print("\nTournament level counts:")
print(df_matches["tourney_level"].value_counts(dropna=False))

print("\nSurface counts:")
print(df_matches["surface"].value_counts(dropna=False))

print("\nRound counts (top 20):")
print(df_matches["round"].value_counts(dropna=False).head(20))

print("\nBest-of counts:")
print(df_matches["best_of"].value_counts(dropna=False))

# -----------------------------------
# 7. Save combined files
# -----------------------------------

print("\n" + "=" * 60)
print("SAVING COMBINED RAW FILES")
print("=" * 60)

df_matches.to_csv(OUTPUT_DIR / "matches_combined_raw.csv", index=False)
df_rankings.to_csv(OUTPUT_DIR / "rankings_combined_raw.csv", index=False)
df_players.to_csv(OUTPUT_DIR / "players_raw.csv", index=False)

print("\nSaved files:")
print(" -", (OUTPUT_DIR / "matches_combined_raw.csv").resolve())
print(" -", (OUTPUT_DIR / "rankings_combined_raw.csv").resolve())
print(" -", (OUTPUT_DIR / "players_raw.csv").resolve())

print("\nDone.")
```
